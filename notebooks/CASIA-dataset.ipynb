```python
import numpy as np
np.bool = np.bool_

import mxnet as mx
from mxnet import recordio

import torch
from torch.utils.data import Subset
from image_iter import FaceDataset, customSubset

import pickle

%matplotlib inline
import matplotlib.pyplot as plt
```

## Create training Dataset and Dataloader

```python
BATCH_SIZE=32
train_root = '/home/pj00/projects/Github/small_face_recognition_trcking/Data/CASIA/casia-webface/train.rec'
```

```python
dataset = FaceDataset(path_imgrec=train_root, rand_mirror=True)

with open('/home/pj00/projects/Github/small_face_recognition_trcking/Data/CASIA/training_dataset.pkl', 'wb') as f:
    pickle.dump(dataset, f)
    
trainloader = torch.utils.data.DataLoader(dataset, batch_size=BATCH_SIZE, shuffle=True, drop_last=True)
torch.save(trainloader, '/home/pj00/projects/Github/small_face_recognition_trcking/Data/CASIA/training_data.pth')
```

```python
ss = customSubset(train_root)
```

```python
len(ss.class_dict.keys())
```

```python
indices = []
for i in range(1000):
    indices += ss.class_dict[i]
```

```python
dataset_sub = Subset(dataset, indices)

with open('/home/pj00/projects/Github/small_face_recognition_trcking/Data/CASIA/training_dataset_sub.pkl', 'wb') as f:
    pickle.dump(dataset_sub, f)
    
trainloader = torch.utils.data.DataLoader(dataset_sub, batch_size=BATCH_SIZE, shuffle=True, drop_last=True)
torch.save(trainloader, '/home/pj00/projects/Github/small_face_recognition_trcking/Data/CASIA/training_data_sub.pth')
```
